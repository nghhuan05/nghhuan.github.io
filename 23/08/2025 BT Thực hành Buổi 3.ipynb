dấu = bình thường, C căn lề phải 5.

```
# Nhập dữ liệu
a = int(input("Mời bạn nhập số a: "))
b = int(input("Mời bạn nhập số b: "))
# Xử lý
kqCong = a + b
kqTru = a - b
kqNhan = a * b
kqChiaNguyen = a // b
kqDu= a % b
kqChiaThuc = a / b
# Xuất dữ liệu
print("Cac ket qua tinh toan: ")
print("%-5d + %5d = %5d"%(a, b, kqCong))
print("%-5d - %5d = %5d"%(a, b, kqTru))
print("%-5d * %5d = %5d"%(a, b, kqNhan))
print("%-5d / %5d = %5d"%(a, b, kqChiaNguyen))
print("%-5d / %5d = %5d"%(a, b, kqDu))
print("%-5d / %5d = %5.2f"%(a, b, kqChiaThuc))
```

**Bài 3: Biểu thức so sánh**

Viết chương trình nhập vào hai số thực a và b. Hãy viết chương trình ước lượng các biểu thức so sánh
A>B, A<B, A>=B, A<=B, A==B, A!=B. Số thực được định dạng có hai số lẻ, có dấu phân cách hàng
ngàn.


```
# Nhập dữ liệu
a = float(input("Moi ban nhap so a: "))
b = float(input("Moi ban nhap so b: "))
# Xử lý
laNhoHon = (a < b)
laNhoHonBang = (a <= b)
laLonHon = (a > b)
laLonHonBang = (a >= b)
laBang = (a == b)
laKhacNhau = (a != b)
# Xuất dữ liệu
print("Ket qua so sanh hai so thuc %.2f va %.2f:"%(a, b))
print("%-8.2f < %8.2f: %8s"%(a, b, laNhoHon))
print("%-8.2f <= %8.2f: %8s"%(a, b, laNhoHonBang))
print("%-8.2f > %8.2f: %8s"%(a, b, laLonHon))
print("%-8.2f >= %8.2f: %8s"%(a, b, laLonHonBang))
print("%-8.2f == %8.2f: %8s"%(a, b, laBang))
print("%-8.2f != %8.2f: %8s"%(a, b, laKhacNhau))
```

**Bài 5: Tính diện tích tam giác**

Cho 3 số thực a, b, c là độ dài 3 cạnh của một tam giác. Hãy tính diện tích tam giác này theo công thức
Heron: 𝑆 = √𝑝(𝑝 − 𝑎)(𝑝 − 𝑏)(𝑝 − 𝑐) với 𝑝 =
𝑎+𝑏+𝑐
2
. In kết quả với 2 số lẻ thập phân

```
a = float(input("Nhap do dai canh a: "))
b = float(input("Nhap do dai canh b: "))
c = float(input("Nhap do dai canh c: "))

p = (a + b + c) / 2

dien_tich = (p * (p - a) * (p - b) * (p - c)) ** 0.5

print("Dien tich tam giac S =", "{:.2f}".format(dien_tich))
```

**Bài 2: Giải phương trình bậc 2**

Nhập 3 số thực a, b, c. Hãy tìm nghiệm của phương trình bậc 2: 𝑎𝑥2+𝑏𝑥+𝑐=0

```
a = float(input("Nhập a: "))
b = float(input("Nhập b: "))
c = float(input("Nhập c: "))

if a == 0:
    if b == 0:
        print("Unlimited solutions")
    else:
        print("No solution")
else:
    delta = b**2 - 4 * a * c
    if delta < 0:
        print("No solution")
    elif delta == 0:
        x = -b / (2 * a)
        print("x1 = x2 =", x)
    else:
        x1 = (-b + (delta ** 0.5)) / (2 * a)
        x2 = (-b - (delta ** 0.5)) / (2 * a)
        print("x1 =", x1, "x2 =", x2)
```

**Bài 3: Tính số ngày của tháng (dùng rẻ nhánh)**

Nhập vào tháng, năm. Hãy cho biết tháng trong năm vừa nhập có bao nhiêu ngày.

```
year = int(input("Nhập năm: "))
month = int(input("Nhập tháng: "))

if month in [4, 6, 9, 11]:
    days = 30
elif month == 2:
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        days = 29  # Năm nhuận
    else:
        days = 28  # Năm không nhuận
else:
    days = 31

print("Tháng", month, "năm", year, "có", days, "ngày")
```
